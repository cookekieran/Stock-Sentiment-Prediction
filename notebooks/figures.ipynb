```python
import matplotlib.pyplot as plt
import pandas as pd

data = {
    'Company': ['Nvidia', 'Microsoft', 'Amazon', 'Apple', 'Alphabet', 'Tesla', 'Meta'],
    'Headlines': [20475, 13735, 8631, 6855, 6084, 5560, 2119]
}

df = pd.DataFrame(data)

plt.figure(figsize=(10, 4))
bars = plt.bar(df['Company'], df['Headlines'], color='#4C72B0', edgecolor='black', zorder=3)

for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2.0, 
        height + 300, 
        f'{height:,}', 
        ha='center', 
        va='bottom', 
        fontsize=10, 
        fontweight='bold'
    )

plt.title('Headlines by MAG7 Company', fontsize=10, fontweight='bold', pad=15)
plt.xlabel('Company', fontsize=10, labelpad=10)
plt.ylabel('Number of Headlines', fontsize=10, labelpad=10)
plt.ylim(0, 24000)
plt.grid(axis='y', linestyle='--', alpha=0.7, zorder=0)

plt.tight_layout()
plt.show()
```

```python
data = {
    'Event Type': [
        'Earnings & guidance',
        'Novel product event',
        'Regulatory / legal outcome',
        'Strategy & capital allocation',
        'Major customer contract',
        'Management/leadership change',
        'Demand/supply revision',
        'Partnership/strategic alliance',
        'M&A/divestiture/asset sale',
        'Cybersecurity/privacy incident',
        'Product safety / quality issue',
        'Supply chain / production disruption'
    ],
    'Count': [3568, 2958, 1462, 1099, 728, 580, 574, 554, 491, 122, 122, 14]
}

df = pd.DataFrame(data)
df = df.sort_values(by='Count', ascending=True)


plt.figure(figsize=(10, 6))
bars = plt.barh(df['Event Type'], df['Count'], color='#2b5c8f', edgecolor='black', zorder=3)

for bar in bars:
    width = bar.get_width()
    plt.text(
        width + 40, 
        bar.get_y() + bar.get_height() / 2.0, 
        f'{width:,}', 
        va='center', 
        ha='left', 
        fontsize=9, 
        fontweight='bold'
    )

plt.title('News Article Counts by Event Type', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Count', fontsize=12, labelpad=10)
plt.ylabel('Event Type', fontsize=12, labelpad=10)
plt.xlim(0, 4100)
plt.grid(axis='x', linestyle='--', alpha=0.7, zorder=0)

plt.tight_layout()
plt.show()
```

```python
data = {
    "Month": [
        "2023-01", "2023-02", "2023-03", "2023-04", "2023-05", "2023-06",
        "2023-07", "2023-08", "2023-09", "2023-10", "2023-11", "2023-12",
        "2024-01", "2024-02", "2024-03", "2024-04", "2024-05", "2024-06",
        "2024-07", "2024-08", "2024-09", "2024-10", "2024-11", "2024-12",
        "2025-01", "2025-02", "2025-03", "2025-04", "2025-05", "2025-06",
        "2025-07", "2025-08", "2025-09", "2025-10", "2025-11", "2025-12",
        "2026-01", "2026-02", "2026-03", "2026-04", "2026-05",
    ],
    "Headlines": [
        347, 340, 290, 174, 206, 228, 184, 222, 208, 182, 210, 214,
        245, 257, 263, 266, 288, 296, 301, 275, 343, 398, 384, 529,
        286, 424, 546, 447, 529, 758, 596, 700, 1387, 10516, 5457,
        4691, 5706, 6276, 6669, 6932, 4889,
    ],
}

df = pd.DataFrame(data)
df["Cumulative Headlines"] = df["Headlines"].cumsum()

plt.figure(figsize=(6, 3))
plt.plot(df["Month"], df["Cumulative Headlines"], color="#1f77b4", linewidth=2)

plt.title("Cumulative Headlines Over Time", fontsize=12)
plt.ylabel("Headlines", fontsize=10)

plt.xticks(df["Month"][::6], rotation=0, fontsize=9)
plt.gca().yaxis.set_major_formatter("{x:,.0f}")

plt.gca().spines["top"].set_visible(False)
plt.gca().spines["right"].set_visible(False)

plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```
